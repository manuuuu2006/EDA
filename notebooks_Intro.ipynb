```python
import numpy as np
arr3 = np.arange(1, 13).reshape(3, 4)
print(arr3)
print(arr3[1, 2])
print(arr3[-1])
print(arr3[:, -1])
```

```python
import numpy as np
arr3 = np.arange(1, 13).reshape(3, 4)
print(arr3[:2, :2])
print(arr3[:, ::2])
print(arr3[:, ::-1])
```

```python
import numpy as np

original_arr = np.arange(1, 21).reshape(4, 5)
view_sub = original_arr[:3, :4]
view_sub[0, 0] = 999
print(original_arr)

copy_sub = original_arr[:3, :4].copy()
copy_sub[0, 0] = 111
print(original_arr)
```
